## 2.  Проверка ошибок в данных и их предобработка


### 2.1. Названия, или метки, столбцов датафрейма


- Выведем на экран названия всех столбцов датафрейма и проверим их стиль написания.


```python
# Выведем на экран названия всех столбцов датафрейма и проверим их стиль написания 
print(df.columns) 
```

- Приведём все столбцы к стилю snake case. 

```python
# Приводим все столбцы к стилю snake case
df.columns = df.columns.str.lower().str.replace(' ', '_')
print(df.columns) 
```

### 2.2. Типы данных



Некоректные типы данны встречаются в столбцах:  EU sales, JP sales и User Score. 
Скорее всего это связано с тем, что цифры были введены как строка. 

Для того, чтобы преобразовать данные типы к типу float64 нужно обработать из этих столбцов пропуски. 

В столбцах EU sales, JP sales пропускаов нет.


- Преобразуем пропуски в столбцах и заменим типы данных с object на float64

```python
df['eu_sales'] = pd.to_numeric(df['eu_sales'], errors='coerce').fillna(0)
df['jp_sales'] = pd.to_numeric(df['jp_sales'], errors='coerce').fillna(0)
df['user_score'] = pd.to_numeric(df['user_score'], errors='coerce').fillna(0)
```

- В числовых столбцах могут встретиться строковые значения, например `unknown` или другие. Приведём такие столбцы к числовому типу данных, заменив строковые значения на пропуски.

```python
df['year_of_release'] = pd.to_numeric(df['year_of_release'], errors='coerce')
df['na_sales'] = pd.to_numeric(df['na_sales'], errors='coerce')
df['other_sales'] = pd.to_numeric(df['other_sales'], errors='coerce')
df['critic_score'] = pd.to_numeric(df['critic_score'], errors='coerce')

```

### 2.3. Наличие пропусков в данных

- Посчитаем количество пропусков в каждом столбце в абсолютных и относительных значениях.


```python
# Посчитаем количество пропусков в каждом столце в абсолютных значениях 
print(df[['name', 'platform', 'year_of_release', 'genre', 'na_sales', 'eu_sales',
       'jp_sales', 'other_sales', 'critic_score', 'user_score', 'rating']].isna().sum())
```

```python
# Посчитаем количество пропусков в каждом столце в относительных значениях 
print(df[['name', 'platform', 'year_of_release', 'genre', 'na_sales', 'eu_sales',
       'jp_sales', 'other_sales', 'critic_score', 'user_score', 'rating']].isna().sum()/len(df))
```

Пропуски характерны для столбцов: 'name', 'year_of_release', 'genre', 'critic_score','rating'.

Наиболшее количество пропусков в столбце 'critic_score' - 8714, это состааляет 51%. Можно предположить, что данный пропуск связан с тем, что не все игры оценивали критики. Такие пропуски лучше предобработать и заменить на какое-либо зачение. 

Также большое колмество пропусков в столбце 'ratin' - 6871, это составляет 40%. Это может быть связано с тем, что не все игры внесены врейтинг организации ESRB. Такие пропуски лучше предобработать и заменить на како-либо зачение. 

Пропуски в столбце 'year_of_release' - 275, это примерно 2%. Эии данные можно либо предоработать, либо проигнорировать, если они не играют роли в анализе. 

В столбцах 'name' и 'genre': 2, это меньше процента. Такие пропуски скорее могут быть связаны с человеческим фактором. Такие данные можно удалить или игнорировать, если они не влияют на результат исследования. 
       

```python
# Удалим пропуски в столбцах: 'name', 'genre'.

df = df.dropna(subset = ['name', 'genre'])
```

```python
# Заменим пропуски в столбце 'year_of_release' средним значением 

df['year_of_release'] = df['year_of_release'].fillna(df['year_of_release'].mean())
```

- Пропуски в столбце 'critic_score' заменять не стоит 

```python
# Заменим пропуски в столбце 'rating' на пустую строку

df['rating'] = df['rating'].fillna('')
```

## 3. Фильтрация данных

Коллеги хотят изучить историю продаж игр в начале XXI века, и их интересует период с 2000 по 2013 год включительно. Отберите данные по этому показателю. Сохраните новый срез данных в отдельном датафрейме, например `df_actual`.

```python
# Создаем новый датафрейм с играми, выпущенными с 2000 по 2013 год включительно

df_actual = df[(df['year_of_release'] >= 2000) & (df['year_of_release'] <= 2013)].copy()
```

```python
# Выодим результат 

print(df_actual.info())
print(df_actual.head())
```

## 4. Категоризация данных
    
Проведите категоризацию данных:
- Разделите все игры по оценкам пользователей и выделите такие категории: высокая оценка (от 8 до 10 включительно), средняя оценка (от 3 до 8, не включая правую границу интервала) и низкая оценка (от 0 до 3, не включая правую границу интервала).

```python
# Создаём пустой список для оценочных групп
evaluation_group_user = []
```

```python
# Классификация игр по оценке пользователей с использованием циклов и условий

for score in df['user_score']:
    
    # Проверяем на NaN (пропущенные значения)
    if pd.isna(score):
        evaluation_group_user.append("нет оценки")
    
    # Проверяем диапазоны оценок (правые границы не включаются, кроме 10)
    elif 8 < score <= 10:  # 10 включается
        evaluation_group_user.append("высокая оценка")
    elif 3 <= score < 8:   # 8 не включается
        evaluation_group_user.append("средняя оценка")
    elif 0 <= score < 3:   # 3 не включается
        evaluation_group_user.append("низкая оценка")
    else:
        evaluation_group_user.append("оценка вне диапазона")
    
```

```python
# Добавляем новый столбец в DataFrame
df['evaluation_group_user'] = evaluation_group_user

print(df['evaluation_group_user'])
```

- Разделим все игры по оценкам критиков и выделим такие категории: высокая оценка (от 80 до 100 включительно), средняя оценка (от 30 до 80, не включая правую границу интервала) и низкая оценка (от 0 до 30, не включая правую границу интервала).

```python
# Создаём пустой список для оценочных групп
evaluation_group_сritic = []
```

```python
# Классификация игр по оценке пользователей с использованием циклов и условий

for score in df['critic_score']:
    
    # Проверяем на NaN (пропущенные значения)
    if pd.isna(score):
        evaluation_group_сritic.append("нет оценки")
    
    # Проверяем диапазоны оценок (правые границы не включаются, кроме 10)
    elif 80 < score <= 100:  # 10 включается
        evaluation_group_сritic.append("высокая оценка")
    elif 30 <= score < 80:   # 8 не включается
        evaluation_group_сritic.append("средняя оценка")
    elif 0 <= score < 30:   # 3 не включается
        evaluation_group_сritic.append("низкая оценка")
    else:
        evaluation_group_сritic.append("оценка вне диапазона")
    
```

```python
# Добавляем новый столбец в DataFrame
df['evaluation_group_сritic'] = evaluation_group_сritic

print(df['evaluation_group_сritic'])
```

- После категоризации данных проверим результат: сгруппируем данные по выделенным категориям и посчитаем количество игр в каждой категории.

```python
# Напишем код для посчета по оценке пользователей 

grouped_1 = df.groupby('evaluation_group_user').agg({'name': 'count'})

print(grouped_1)
```

```python
# Напишем код для посчета по оценке критиков 

grouped_2 = df.groupby('evaluation_group_сritic').agg({'name': 'count'})

print(grouped_2)
```

- Выделим топ-7 платформ по количеству игр, выпущенных за весь актуальный период.

```python
# Сгруппируем данные по названию платформы и посчитаем количество игр на каждой платформе

df['game_count'] = df.groupby('platform').agg({'name': 'count'})
```

```python
# Отсортируем по колисевту игр на платформе в порядке убывания и выведем 7 первых значений.

df_sorted = df.sort_values(by='game_count', ascending=False)
print(df_sorted.head(7))
```
